```python
def get_NPS_scores(df_quant_admin, df_quant_faculty, df_quant_student) -> pd.DataFrame:
    avg_NPS_admin = df_quant_admin['dashboard_use_nps_score'].mean()
    avg_NPS_faculty = df_quant_faculty['library_partner_nps_score'].mean()
    avg_NPS_student = df_quant_student['library_nps_score'].mean()
    NPS_scores = {'group':['Students','Faculty','Admin'], 'scores':[avg_NPS_student,avg_NPS_faculty,avg_NPS_admin]}
    df = pd.DataFrame.from_dict(NPS_scores)
    return df
```

```python
def get_impact_satisfaction_ratings(df_admin, df_faculty, df_student) -> pd.DataFrame:
    data_support_effectiveness = df_admin['data_support_effectiveness'].dropna().mean()
    health_info_confidence = df_student['health_info_confidence'].dropna().mean()
    collection_support = df_faculty['collection_support_rating'].dropna().mean()
    research_materials_satisfaction = df_faculty['research_materials_satisfaction'].dropna().mean()
    course_materials_satisfaction = df_faculty['course_materials_satisfaction'].dropna().mean()
    satisfaction_ratings = {'rating':['Data Support Effectiveness','Health Info Confidence','Collection Support', 'Research Materials Satisfaction', 'Course Materials Satisfaction'], 'scores':[data_support_effectiveness,health_info_confidence,collection_support, research_materials_satisfaction, course_materials_satisfaction]}
    satisfaction_ratings_df = pd.DataFrame.from_dict(satisfaction_ratings)
    return satisfaction_ratings_df
```

```python
satisfaction_ratings_df = get_impact_satisfaction_ratings(df_admin, df_faculty, df_student)
```
